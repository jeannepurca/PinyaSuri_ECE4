<h2>Import libraries:</h2>

```python
import pandas as pd
import numpy as np
from math import radians, sin, cos, sqrt, asin
```

<h2>Load raw telemetry CSV:</h2>

```python
# Load raw telemetry CSV
RAW_CSV = "flight_raw_data.csv"
df = pd.read_csv(RAW_CSV)

# Convert timestamp to datetime
df['timestamp_utc'] = pd.to_datetime(df['timestamp_utc'])
df.head()
```

<h2>Group data by flight number:</h2>

```python
df = df.sort_values("timestamp_utc").reset_index(drop=True)

df["time_diff_s"] = df["timestamp_utc"].diff().dt.total_seconds()
```

<h2>Haversine function (GPS distance):</h2>

```python
def haversine(lat1, lon1, lat2, lon2):
    # Compute horizontal distance (meters) between two lat/lon points
    R = 6371000  # Earth radius (meters)
    
    phi1 = radians(lat1)
    phi2 = radians(lat2)
    dphi = radians(lat2 - lat1)
    dlambda = radians(lon2 - lon1)

    a = sin(dphi/2)**2 + cos(phi1)*cos(phi2)*sin(dlambda/2)**2
    return 2 * R * asin(sqrt(a))
```

<h2>Compute metrics per flight:</h2>

```python
results = []

for flight_id, flight_df in grouped:
    metrics = {"flight_number": flight_id}

    # --------------------------------
    # Altitude / Attitude Stability
    # --------------------------------
    metrics["roll_std_deg"] = flight_df["roll_deg"].std()
    metrics["pitch_std_deg"] = flight_df["pitch_deg"].std()
    metrics["yaw_std_deg"] = flight_df["yaw_deg"].std()

    metrics["altitude_rms_m"] = np.sqrt(
        np.mean((flight_df["alt_m"] - flight_df["alt_m"].mean())**2)
    )

    # --------------------------------
    # IMU Vibration (Acceleration RMS)
    # --------------------------------
    accel_mag = np.sqrt(
        flight_df["accel_x_m_s2"]**2 +
        flight_df["accel_y_m_s2"]**2 +
        flight_df["accel_z_m_s2"]**2
    )
    metrics["imu_accel_rms_m_s2"] = np.sqrt(np.mean(accel_mag**2))

    # --------------------------------
    # Waypoint Navigation Accuracy
    # --------------------------------
    wp_distances = []

    for _, row in flight_df.iterrows():
        d_hor = haversine(
            row["lat_deg"], row["lon_deg"],
            row["waypoint_lat_deg"], row["waypoint_lon_deg"]
        )
        d_vert = abs(row["alt_m"] - row["waypoint_alt_m"])
        wp_distances.append(sqrt(d_hor**2 + d_vert**2))

    metrics["waypoint_mean_error_m"] = np.mean(wp_distances)
    metrics["waypoint_rms_error_m"] = np.sqrt(
        np.mean(np.array(wp_distances)**2)
    )

    # --------------------------------
    # Position Jitter During Hover
    # --------------------------------
    hover_df = flight_df[flight_df["is_hovering"] == True]

    if not hover_df.empty:
        lat0 = hover_df["lat_deg"].mean()
        lon0 = hover_df["lon_deg"].mean()
        alt0 = hover_df["alt_m"].mean()

        jitter_dist = hover_df.apply(
            lambda r: sqrt(
                haversine(r["lat_deg"], r["lon_deg"], lat0, lon0)**2 +
                (r["alt_m"] - alt0)**2
            ),
            axis=1
        )

        metrics["hover_jitter_rms_m"] = np.sqrt(np.mean(jitter_dist**2))
    else:
        metrics["hover_jitter_rms_m"] = np.nan

    # --------------------------------
    # Flight Endurance
    # --------------------------------
    start_time = flight_df["timestamp_utc"].min()
    end_time = flight_df["timestamp_utc"].max()

    metrics["flight_time_s"] = (end_time - start_time).total_seconds()
    metrics["avg_battery_current_A"] = flight_df["battery_current_A"].mean()
    metrics["avg_battery_voltage_V"] = flight_df["battery_voltage_V"].mean()

    metrics["battery_used_percent"] = (
        flight_df["battery_percentage"].iloc[0] -
        flight_df["battery_percentage"].iloc[-1]
    )

    results.append(metrics)
```

<h2>Create final per-flight summary table:</h2>

```python
summary_df = pd.DataFrame(results)
summary_df
```

<h2>Save results to CSV:</h2>

```python
summary_df.to_csv("flight_metrics_summary.csv", index=False)
```
