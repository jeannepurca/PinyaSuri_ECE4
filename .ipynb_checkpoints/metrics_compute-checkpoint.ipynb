# Computes flight performance metrics from raw telemetry data

# IMPORT LIBRARIES

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from math import radians, sin, cos, sqrt, asin
from pathlib import Path
import warnings

warnings.filterwarnings("ignore")
```

# FILE PATHS & CONSTANTS

```python
RAW_CSV = Path("raw_flight_data.csv")
SUMMARY_CSV = Path("flight_metrics_summary.csv")
OVERALL_CSV = Path("overall_metrics.csv")
```

# HAVERSINE DISTANCE

```python
EARTH_RADIUS_M = 6371000  # meters

def haversine(lat1, lon1, lat2, lon2):
    """Calculate great-circle distance between two GPS coordinates"""
    lat1, lon1, lat2, lon2 = map(radians, [lat1, lon1, lat2, lon2])
    dlat = lat2 - lat1
    dlon = lon2 - lon1
    a = sin(dlat / 2) ** 2 + cos(lat1) * cos(lat2) * sin(dlon / 2) ** 2
    return 2 * EARTH_RADIUS_M * asin(sqrt(a))
```

# LOAD DATA / CSV FILE

```python
try:
    df = pd.read_csv(
        RAW_CSV,
        engine="python",
        on_bad_lines="skip"
    )
    df["timestamp_utc"] = pd.to_datetime(df["timestamp_utc"])
    print(f"✓ Loaded {len(df)} telemetry records")
    print(f"✓ Columns: {list(df.columns)}")
except FileNotFoundError:
    raise SystemExit(f"❌ Error: {RAW_CSV} not found")
except Exception as e:
    raise SystemExit(f"❌ Error loading data: {e}")
```

# DATA OVERVIEW

```python
flights = df.groupby("flight_id")
print(f"\nTotal flights in dataset: {len(flights)}")
for flight_id, group in flights:
    print(f"  - {flight_id}: {len(group)} records")
```

# METRIC THRESHOLDS

```python
THRESHOLDS = {
    "altitude_stability_index": {"good": 5.0, "acceptable": 10.0},
    "imu_vibration_rms_m_s2": {"good": 8.0, "acceptable": 12.0},
    "waypoint_accuracy_rms_m": {"good": 1.5, "acceptable": 3.0},
    "hover_position_jitter_rms_m": {"good": 1.0, "acceptable": 2.5},
    "avg_flight_duration_s": {"good": 60.0, "acceptable": 30.0},
}
```

# JUDGE METRIC

```python
def judge_metric(value, good, acceptable, higher_is_better=False):
    if np.isnan(value):
        return "N/A"
    if higher_is_better:
        return "Good" if value >= good else "Acceptable" if value >= acceptable else "Poor"
    return "Good" if value <= good else "Acceptable" if value <= acceptable else "Poor"
```

# PER FLIGHT COMPUTATION

```python
results = []

for flight_id, f in flights:
    roll_std = f["roll_deg"].std()
    pitch_std = f["pitch_deg"].std()
    yaw_std = f["yaw_deg"].std()
    alt_std = f["alt_m"].std()

    accel_mag = np.sqrt(
        f["accel_x_m_s2"]**2 +
        f["accel_y_m_s2"]**2 +
        f["accel_z_m_s2"]**2
    )
    imu_rms = np.sqrt(np.mean(accel_mag**2))

    wp = f[f["waypoint_index"] >= 0].copy()
    if not wp.empty:
        wp["error_m"] = wp.apply(
            lambda r: haversine(
                r["lat_deg"], r["lon_deg"],
                r["waypoint_lat_deg"], r["waypoint_lon_deg"]
            )
            if pd.notna(r["waypoint_lat_deg"]) else np.nan,
            axis=1
        )
        wp_mean = wp["error_m"].mean()
        wp_rms = np.sqrt(np.mean(wp["error_m"]**2))
    else:
        wp_mean = wp_rms = np.nan

    hover = f[f["is_hovering"] == True]
    if len(hover) > 1:
        lat_m = hover["lat_deg"].mean()
        lon_m = hover["lon_deg"].mean()
        jitter = hover.apply(
            lambda r: haversine(r["lat_deg"], r["lon_deg"], lat_m, lon_m),
            axis=1
        )
        hover_rms = np.sqrt(np.mean(jitter**2))
    else:
        hover_rms = np.nan

    duration = (f["timestamp_utc"].iloc[-1] - f["timestamp_utc"].iloc[0]).total_seconds()

    results.append({
        "flight_id": flight_id,
        "duration_s": round(duration, 1),
        "roll_std_deg": round(roll_std, 3),
        "pitch_std_deg": round(pitch_std, 3),
        "yaw_std_deg": round(yaw_std, 3),
        "altitude_std_m": round(alt_std, 3),
        "imu_accel_rms_m_s2": round(imu_rms, 3),
        "wp_error_mean_m": round(wp_mean, 2),
        "wp_error_rms_m": round(wp_rms, 2),
        "hover_jitter_rms_m": round(hover_rms, 2),
    })

summary_df = pd.DataFrame(results)
summary_df.to_csv(SUMMARY_CSV, index=False)
print(f"\n✓ Per-flight metrics saved to {SUMMARY_CSV}")
```

# SAVE SUMMARY

```python
summary_df = pd.DataFrame(results)
```

# OVERALL METRICS

```python
overall = {}

alt_stability = np.nanmean([
    summary_df["roll_std_deg"].mean(),
    summary_df["pitch_std_deg"].mean(),
    summary_df["yaw_std_deg"].mean(),
    summary_df["altitude_std_m"].mean()
])

overall["altitude_stability_index"] = round(alt_stability, 3)
overall["altitude_stability_rating"] = judge_metric(
    alt_stability,
    THRESHOLDS["altitude_stability_index"]["good"],
    THRESHOLDS["altitude_stability_index"]["acceptable"]
)

imu_avg = summary_df["imu_accel_rms_m_s2"].mean()
overall["imu_vibration_rms_m_s2"] = round(imu_avg, 3)
overall["imu_vibration_rating"] = judge_metric(
    imu_avg,
    THRESHOLDS["imu_vibration_rms_m_s2"]["good"],
    THRESHOLDS["imu_vibration_rms_m_s2"]["acceptable"]
)

wp_avg = summary_df["wp_error_rms_m"].dropna().mean()
overall["waypoint_accuracy_rms_m"] = round(wp_avg, 2)
overall["waypoint_accuracy_rating"] = judge_metric(
    wp_avg,
    THRESHOLDS["waypoint_accuracy_rms_m"]["good"],
    THRESHOLDS["waypoint_accuracy_rms_m"]["acceptable"]
)

hover_avg = summary_df["hover_jitter_rms_m"].dropna().mean()
overall["hover_position_jitter_rms_m"] = round(hover_avg, 2)
overall["hover_position_jitter_rating"] = judge_metric(
    hover_avg,
    THRESHOLDS["hover_position_jitter_rms_m"]["good"],
    THRESHOLDS["hover_position_jitter_rms_m"]["acceptable"]
)

duration_avg = summary_df["duration_s"].mean()
overall["avg_flight_duration_s"] = round(duration_avg, 1)
overall["flight_endurance_rating"] = judge_metric(
    duration_avg,
    THRESHOLDS["avg_flight_duration_s"]["good"],
    THRESHOLDS["avg_flight_duration_s"]["acceptable"],
    higher_is_better=True
)

overall_df = pd.DataFrame(overall, index=["overall"])
overall_df.to_csv(OVERALL_CSV)
print(f"✓ Overall metrics saved to {OVERALL_CSV}")

RATING_MAP = {
    "altitude_stability_index": "altitude_stability_rating",
    "imu_vibration_rms_m_s2": "imu_vibration_rating",
    "waypoint_accuracy_rms_m": "waypoint_accuracy_rating",
    "hover_position_jitter_rms_m": "hover_position_jitter_rating",
    "avg_flight_duration_s": "flight_endurance_rating",
}

print("\n" + "=" * 70)
print("FLIGHT PERFORMANCE SUMMARY")
print("=" * 70)

for metric, rating_key in RATING_MAP.items():
    value = overall.get(metric, np.nan)
    rating = overall.get(rating_key, "N/A")
    print(f"{metric:40s}: {value:10.3f}  [{rating}]")

print("=" * 70)
```

# 4. Hover Position Jitter

```python
# XY Scatter During Hover

hover_df = df[df["flight_mode"] == "HOVER"]

plt.figure()
plt.scatter(
    hover_df["longitude"],
    hover_df["latitude"],
    s=5
)
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("Hover Position Jitter")
plt.grid(True)
plt.show()
```

```python
# Distance-from-Mean Jitter

mean_lat = hover_df["latitude"].mean()
mean_lon = hover_df["longitude"].mean()

hover_df["jitter_m"] = hover_df.apply(
    lambda r: haversine(
        r["latitude"], r["longitude"],
        mean_lat, mean_lon
    ),
    axis=1
)

plt.figure()
plt.hist(hover_df["jitter_m"], bins=30)
plt.xlabel("Jitter Distance (m)")
plt.ylabel("Frequency")
plt.title("Hover Position Jitter Distribution")
plt.grid(True)
plt.show()
```
